```python
import pandas as pd

data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Math': [85, 90, 95],
    'Science': [80, 85, 90]
}

df = pd.DataFrame(data)
print(df)


#melted makes it a longer format
melted_df = pd.melt(df, id_vars=['Name'], value_vars=['Math', 'Science'],
                    var_name='Subject', value_name='Score')

print(melted_df)
```
